```python
# %% [markdown]
# # Classification of rice plant diseases using DenseNet121 — replication pipeline
# Paper: Ismail et al., Scientific Reports 16:7482 (2026)
# Data : https://www.kaggle.com/competitions/paddy-disease-classification/data
#
# Pipeline (as described in the paper)
#   1. Load Paddy data, keep the 7 disease classes
#   2. Stratified 80/20 train/test split, real-time augmentation
#      (rotation, zoom, horizontal+vertical flip, brightness)
#   3. DenseNet121 (ImageNet) + custom head: GAP -> Dropout(0.5) -> Dense(256, ReLU)
#      -> Dropout(0.3) -> Dense(N, softmax)
#   4. Phase 1: freeze dense blocks 1-3, train block 4 + head (Adam, lr=1e-3)
#      Phase 2: unfreeze blocks 3-4, fine-tune (Adam, lr=1e-4)
#   5. Categorical cross-entropy, batch 32, <=50 epochs, ReduceLROnPlateau(0.1, pat 3),
#      EarlyStopping(pat 5) on val_loss
#   6. Metrics: sensitivity, specificity, precision, F-score, accuracy (Eq. 1-4),
#      Table 4 / Table 5, Fig. 4 (curves), Fig. 5 (row-normalised confusion matrix)
#   7. 5-fold stratified CV (Table 7), 95% CI + significance test (Table 8)

# %% 1. Imports and configuration
import os
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.applications.densenet import preprocess_input
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, precision_recall_fscore_support, accuracy_score
from scipy import stats

# ---- reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
keras.utils.set_random_seed(SEED)

# ---- data
# The paper uses 7 disease classes. Kaggle's Paddy set has 10 (adds dead_heart, hispa, normal).
# Set KEEP_CLASSES = None to use all 10.
KEEP_CLASSES = [
    "bacterial_leaf_blight",
    "bacterial_leaf_streak",
    "bacterial_panicle_blight",
    "blast",
    "brown_spot",
    "downy_mildew",
    "tungro",
]
IMG_SIZE = (224, 224)          # paper does not state the input size; DenseNet121 default
TEST_SIZE = 0.20               # text says 80/20 (note: Table 3 actually shows ~70/30)
VAL_FRAC_OF_TRAIN = 0.10       # validation carved out of the training split (used by callbacks)
USE_TEST_AS_VAL = False        # True = what the paper's curves appear to do (test used as val)

# ---- training
BATCH_SIZE = 32
PHASE1_EPOCHS = 20             # frozen blocks 1-3
PHASE2_EPOCHS = 30             # fine-tune blocks 3-4  (20 + 30 = 50 max epochs)
LR_PHASE1 = 1e-3
LR_PHASE2 = 1e-4
DROPOUT_1, DROPOUT_2, DENSE_UNITS = 0.5, 0.3, 256
LR_FACTOR, LR_PATIENCE = 0.1, 3
ES_PATIENCE = 5

# ---- experiments
RUN_MAIN = True
RUN_CV = False                 # 5 full trainings, expensive. Turn on when the main run looks right.
N_FOLDS = 5
BASELINE_PCT = 94.0            # Saputra et al. DenseNet121 (Table 6), reference for the t-test

# ---- paths
OUT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)


def find_data_root():
    candidates = [
        "/kaggle/input/paddy-disease-classification",
        "/content/paddy-disease-classification",
        "./paddy-disease-classification",
    ]
    for c in candidates:
        if (Path(c) / "train.csv").exists():
            return Path(c)
    # Colab fallback: needs ~/.kaggle/kaggle.json and the competition rules accepted on the website
    os.system("pip -q install kaggle")
    os.system("kaggle competitions download -c paddy-disease-classification -p /content")
    os.system("unzip -q -o /content/paddy-disease-classification.zip -d /content/paddy-disease-classification")
    return Path("/content/paddy-disease-classification")


DATA_ROOT = find_data_root()
print("Data root:", DATA_ROOT)
print("GPUs:", tf.config.list_physical_devices("GPU"))

# %% 2. Load metadata and build the dataframe
# train.csv columns: image_id, label, variety, age. Images: train_images/<label>/<image_id>
# (test_images has no labels, so everything is split from train_images.)
df = pd.read_csv(DATA_ROOT / "train.csv")
if KEEP_CLASSES is not None:
    df = df[df["label"].isin(KEEP_CLASSES)].reset_index(drop=True)

CLASSES = sorted(df["label"].unique())
NUM_CLASSES = len(CLASSES)
label2idx = {c: i for i, c in enumerate(CLASSES)}
df["y"] = df["label"].map(label2idx)
df["path"] = [str(DATA_ROOT / "train_images" / l / i) for i, l in zip(df["image_id"], df["label"])]
PRETTY = [c.replace("_", " ").capitalize() for c in CLASSES]

print(f"{len(df)} images, {NUM_CLASSES} classes")
print(df["label"].value_counts().rename("n_images"))

# %% 3. Splits
def split_train_val(train_df):
    tr, va = train_test_split(
        train_df, test_size=VAL_FRAC_OF_TRAIN, stratify=train_df["y"], random_state=SEED
    )
    return tr.reset_index(drop=True), va.reset_index(drop=True)


train_df, test_df = train_test_split(df, test_size=TEST_SIZE, stratify=df["y"], random_state=SEED)
train_df, test_df = train_df.reset_index(drop=True), test_df.reset_index(drop=True)

if USE_TEST_AS_VAL:
    val_df = test_df
else:
    train_df, val_df = split_train_val(train_df)

print(f"train={len(train_df)}  val={len(val_df)}  test={len(test_df)}")

# %% 4. tf.data input pipeline with real-time augmentation
AUTOTUNE = tf.data.AUTOTUNE

augment = keras.Sequential(
    [
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(20 / 360),                       # about +/-20 degrees
        layers.RandomZoom(0.2),
        layers.RandomBrightness(0.2, value_range=(0, 255)),
    ],
    name="augment",
)


def _load(path, y):
    img = tf.io.read_file(path)
    img = tf.io.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    return img, tf.one_hot(y, NUM_CLASSES)


def make_ds(frame, training):
    ds = tf.data.Dataset.from_tensor_slices((frame["path"].values, frame["y"].values))
    if training:
        ds = ds.shuffle(len(frame), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(_load, num_parallel_calls=AUTOTUNE)
    if training:
        ds = ds.map(lambda x, y: (augment(x, training=True), y), num_parallel_calls=AUTOTUNE)
    ds = ds.map(lambda x, y: (preprocess_input(x), y), num_parallel_calls=AUTOTUNE)
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)


# %% 5. Model: DenseNet121 + custom head, with block-wise freezing
# Keras DenseNet121 layer naming:
#   conv1_*            stem
#   conv2_block*       dense block 1 (6 layers)     pool2_*  transition 1
#   conv3_block*       dense block 2 (12 layers)    pool3_*  transition 2
#   conv4_block*       dense block 3 (24 layers)    pool4_*  transition 3
#   conv5_block*       dense block 4 (16 layers)
#   bn                 final batch norm
PHASE1_TRAINABLE = ("conv5_", "bn")                       # block 4 + final BN (+ head)
PHASE2_TRAINABLE = ("conv4_", "pool4_", "conv5_", "bn")   # blocks 3 and 4 (+ head)


def build_model():
    base = DenseNet121(include_top=False, weights="imagenet", input_shape=(*IMG_SIZE, 3))
    x = layers.GlobalAveragePooling2D(name="gap")(base.output)
    x = layers.Dropout(DROPOUT_1, name="drop1")(x)
    x = layers.Dense(DENSE_UNITS, activation="relu", kernel_initializer="glorot_uniform", name="fc256")(x)
    x = layers.Dropout(DROPOUT_2, name="drop2")(x)
    out = layers.Dense(NUM_CLASSES, activation="softmax", kernel_initializer="glorot_uniform", name="out")(x)
    model = keras.Model(base.input, out, name="densenet121_rice")
    base_names = {l.name for l in base.layers}
    return model, base_names


def set_trainable(model, base_names, prefixes):
    for l in model.layers:
        if l.name in base_names:
            l.trainable = l.name.startswith(prefixes)
        else:
            l.trainable = True  # classification head


def compile_model(model, lr):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss="categorical_crossentropy",
        metrics=["accuracy"],
    )


def make_callbacks():
    return [
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=LR_FACTOR, patience=LR_PATIENCE, min_lr=1e-7, verbose=1
        ),
        keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=ES_PATIENCE, restore_best_weights=True, verbose=1
        ),
    ]


def merge_history(h1, h2):
    return {k: h1.history[k] + h2.history[k] for k in h1.history}


def train_two_phase(train_ds, val_ds, verbose=1):
    model, base_names = build_model()

    # Phase 1 - frozen blocks 1-3, lr = 1e-3
    set_trainable(model, base_names, PHASE1_TRAINABLE)
    compile_model(model, LR_PHASE1)
    h1 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE1_EPOCHS,
                   callbacks=make_callbacks(), verbose=verbose)

    # Phase 2 - unfreeze blocks 3 and 4, lr = 1e-4
    set_trainable(model, base_names, PHASE2_TRAINABLE)
    compile_model(model, LR_PHASE2)
    h2 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE2_EPOCHS,
                   callbacks=make_callbacks(), verbose=verbose)

    return model, merge_history(h1, h2)


# %% 6. Metrics (Eq. 1-4 of the paper, one-vs-rest per class)
def per_class_metrics(cm):
    cm = cm.astype(np.float64)
    tp = np.diag(cm)
    fn = cm.sum(axis=1) - tp
    fp = cm.sum(axis=0) - tp
    tn = cm.sum() - tp - fn - fp
    eps = 1e-12
    sens = tp / (tp + fn + eps)                        # Eq. 1 (recall)
    spec = tn / (tn + fp + eps)                        # Eq. 2
    acc = (tp + tn) / (tp + tn + fp + fn + eps)        # Eq. 3 (per-class, one-vs-rest)
    prec = tp / (tp + fp + eps)                        # Eq. 4
    f1 = 2 * prec * sens / (prec + sens + eps)
    return sens, spec, prec, f1, acc


def evaluate(model, frame):
    ds = make_ds(frame, training=False)          # not shuffled, order matches frame
    y_true = frame["y"].values
    y_pred = model.predict(ds, verbose=0).argmax(axis=1)
    cm = confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES)))

    sens, spec, prec, f1, acc = per_class_metrics(cm)
    table4 = pd.DataFrame(
        {
            "Sensitivity": sens * 100,
            "Specificity": spec * 100,
            "Precision": prec * 100,
            "F-score": f1 * 100,
            "Accuracy": acc * 100,
        },
        index=PRETTY,
    ).round(2)
    table4.index.name = "Disease Types"

    overall_acc = np.trace(cm) / cm.sum() * 100
    table5 = pd.DataFrame(
        {
            "Sensitivity": [sens.mean() * 100],
            "Specificity": [spec.mean() * 100],
            "Precision": [prec.mean() * 100],
            "F-score": [f1.mean() * 100],
            "Accuracy": [overall_acc],
        },
        index=["Our proposed model"],
    ).round(2)
    return cm, table4, table5, y_true, y_pred


# %% 7. Plots (Fig. 4 and Fig. 5)
def plot_curves(hist, path):
    ep = np.arange(1, len(hist["loss"]) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
    ax[0].plot(ep, hist["loss"], "o-", ms=3, label="Training Loss")
    ax[0].plot(ep, hist["val_loss"], "x-", ms=4, label="Validation Loss")
    ax[0].set(title="Training and Validation Loss", xlabel="Epochs", ylabel="Loss")
    ax[1].plot(ep, hist["accuracy"], "o-", ms=3, label="Training Accuracy")
    ax[1].plot(ep, hist["val_accuracy"], "x-", ms=4, label="Validation Accuracy")
    ax[1].set(title="Training and Validation Accuracy", xlabel="Epochs", ylabel="Accuracy")
    for a in ax:
        a.grid(alpha=0.3)
        a.legend()
    plt.tight_layout()
    plt.savefig(path, dpi=300)
    plt.show()


def plot_confusion(cm, path):
    cmn = cm / cm.sum(axis=1, keepdims=True)
    plt.figure(figsize=(9, 7))
    sns.heatmap(
        cmn, annot=True, fmt=".2%", cmap="Blues", vmin=0, vmax=1,
        xticklabels=PRETTY, yticklabels=PRETTY, cbar_kws={"label": "Percentage"},
    )
    plt.title("Confusion Matrix for Rice Disease Classification (Normalized by Row)")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(path, dpi=300)
    plt.show()


# %% 8. Main experiment: train once, evaluate on the held-out test split
if RUN_MAIN:
    train_ds = make_ds(train_df, training=True)
    val_ds = make_ds(val_df, training=False)

    model, history = train_two_phase(train_ds, val_ds)
    model.save(OUT_DIR / "densenet121_rice.keras")
    pd.DataFrame(history).to_csv(OUT_DIR / "history.csv", index=False)

    cm, table4, table5, y_true, y_pred = evaluate(model, test_df)

    print("\nTable 4 - per-class results (%)")
    print(table4)
    print("\nTable 5 - overall results (%)")
    print(table5)

    table4.to_csv(OUT_DIR / "table4_per_class.csv")
    table5.to_csv(OUT_DIR / "table5_overall.csv")
    np.save(OUT_DIR / "confusion_matrix.npy", cm)

    plot_curves(history, OUT_DIR / "fig4_curves.png")
    plot_confusion(cm, OUT_DIR / "fig5_confusion_matrix.png")

# %% 9. 5-fold cross-validation (Table 7) and 95% CI / significance (Table 8)
def run_cross_validation():
    skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
    rows = []
    for k, (tr_idx, te_idx) in enumerate(skf.split(df, df["y"]), start=1):
        print(f"\n===== Fold {k}/{N_FOLDS} =====")
        keras.backend.clear_session()
        tr_full = df.iloc[tr_idx].reset_index(drop=True)
        te = df.iloc[te_idx].reset_index(drop=True)
        tr, va = split_train_val(tr_full)

        m, _ = train_two_phase(make_ds(tr, True), make_ds(va, False), verbose=2)

        y_pred = m.predict(make_ds(te, False), verbose=0).argmax(axis=1)
        y_true = te["y"].values
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)
        rows.append(
            {
                "Fold": f"Fold {k}",
                "Accuracy (%)": accuracy_score(y_true, y_pred) * 100,
                "Precision (%)": p * 100,
                "Recall (%)": r * 100,
                "F1-Score (%)": f * 100,
            }
        )
        print(rows[-1])
        del m

    cv = pd.DataFrame(rows).set_index("Fold")
    metrics = list(cv.columns)

    # Table 7: mean +/- SD (sample SD, ddof=1)
    mean, sd = cv.mean(), cv.std(ddof=1)
    table7 = cv.round(2).copy()
    table7.loc["Mean ± SD"] = [f"{mean[c]:.1f} ± {sd[c]:.2f}" for c in metrics]
    print("\nTable 7 - 5-fold cross-validation")
    print(table7)

    # Table 8: t-based 95% CI over folds + one-sample t-test against a reference baseline
    n = len(cv)
    tcrit = stats.t.ppf(0.975, df=n - 1)
    t8 = []
    for c in metrics:
        half = tcrit * sd[c] / np.sqrt(n)
        _, p = stats.ttest_1samp(cv[c].values, BASELINE_PCT)
        t8.append(
            {
                "Metric": c.replace(" (%)", ""),
                "Mean (%)": round(mean[c], 1),
                "95% CI": f"[{mean[c] - half:.1f}, {mean[c] + half:.1f}]",
                "p-value": "<0.001" if p < 0.001 else f"{p:.3f}",
            }
        )
    table8 = pd.DataFrame(t8).set_index("Metric")
    print(f"\nTable 8 - 95% CI and significance (vs. {BASELINE_PCT}% baseline)")
    print(table8)

    table7.to_csv(OUT_DIR / "table7_cv.csv")
    table8.to_csv(OUT_DIR / "table8_ci.csv")
    return cv, table7, table8


if RUN_CV:
    cv_raw, table7, table8 = run_cross_validation()
```

```python
import os
import random
from pathlib import Path

import matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.applications.densenet import preprocess_input
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, precision_recall_fscore_support, accuracy_score
from sklearn.utils.class_weight import compute_class_weight
from scipy import stats

# ---- reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
keras.utils.set_random_seed(SEED)

# ---- data
# The paper uses 7 disease classes. Kaggle's Paddy set has 10 (adds dead_heart, hispa, normal).
# Set KEEP_CLASSES = None to use all 10.
KEEP_CLASSES = [
    "bacterial_leaf_blight",
    "bacterial_leaf_streak",
    "bacterial_panicle_blight",
    "blast",
    "brown_spot",
    "downy_mildew",
    "tungro",
]
# IMPROVEMENT 1: higher resolution, keeping the 3:4 aspect ratio of the original photos
# (originals are 480x640; squashing to 224x224 destroys small lesion detail).
# DenseNet121 + GlobalAveragePooling accepts any size. If you hit out-of-memory, use (256, 320).
IMG_SIZE = (288, 384)          # (height, width)
TEST_SIZE = 0.20
VAL_FRAC_OF_TRAIN = 0.10       # validation carved out of the training split (used by callbacks)
USE_TEST_AS_VAL = False        # keep False: using the test set for callbacks leaks information

# ---- training
BATCH_SIZE = 32                # lower to 16 if the GPU runs out of memory
PHASE1_EPOCHS = 8              # warm-up: new head only (backbone frozen)
PHASE2_EPOCHS = 45             # fine-tune the whole backbone (early stopping decides the real length)
LR_PHASE1 = 1e-3
LR_PHASE2 = 1e-4               # whole network at 1e-4, reduced on plateau down to 1e-7
DROPOUT_1, DROPOUT_2, DENSE_UNITS = 0.5, 0.3, 256
LR_FACTOR, LR_PATIENCE = 0.3, 2
ES_PATIENCE = 8
LABEL_SMOOTHING = 0.1          # IMPROVEMENT 2: reduces over-confidence, usually +0.5-1% on noisy field images
USE_CLASS_WEIGHTS = True       # IMPROVEMENT 3: dataset is imbalanced (blast ~1700 vs panicle blight ~340)
USE_TTA = True                 # IMPROVEMENT 4: flip test-time augmentation at evaluation
USE_MIXED_PRECISION = True     # faster and lighter on memory (needs a GPU)

# ---- experiments
RUN_MAIN = True
RUN_CV = False                 # 5 full trainings, expensive. Turn on when the main run looks right.
N_FOLDS = 5
BASELINE_PCT = 94.0            # Saputra et al. DenseNet121 (Table 6), reference for the t-test

if USE_MIXED_PRECISION and tf.config.list_physical_devices("GPU"):
    keras.mixed_precision.set_global_policy("mixed_float16")

# ---- paths
OUT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)


def find_data_root():
    candidates = [
        "/kaggle/input/paddy-disease-classification",
        "/content/paddy-disease-classification",
        "./paddy-disease-classification",
    ]
    for c in candidates:
        if (Path(c) / "train.csv").exists():
            return Path(c)
    # Colab fallback: needs ~/.kaggle/kaggle.json and the competition rules accepted on the website
    os.system("pip -q install kaggle")
    os.system("kaggle competitions download -c paddy-disease-classification -p /content")
    os.system("unzip -q -o /content/paddy-disease-classification.zip -d /content/paddy-disease-classification")
    return Path("/content/paddy-disease-classification")


DATA_ROOT = find_data_root()
print("Data root:", DATA_ROOT)
print("GPUs:", tf.config.list_physical_devices("GPU"))

# %% 2. Load metadata and build the dataframe
# train.csv columns: image_id, label, variety, age. Images: train_images/<label>/<image_id>
# (test_images has no labels, so everything is split from train_images.)
df = pd.read_csv(DATA_ROOT / "train.csv")
if KEEP_CLASSES is not None:
    df = df[df["label"].isin(KEEP_CLASSES)].reset_index(drop=True)

CLASSES = sorted(df["label"].unique())
NUM_CLASSES = len(CLASSES)
label2idx = {c: i for i, c in enumerate(CLASSES)}
df["y"] = df["label"].map(label2idx)
df["path"] = [str(DATA_ROOT / "train_images" / l / i) for i, l in zip(df["image_id"], df["label"])]
PRETTY = [c.replace("_", " ").capitalize() for c in CLASSES]

print(f"{len(df)} images, {NUM_CLASSES} classes")
print(df["label"].value_counts().rename("n_images"))


# %% 3. Splits
def split_train_val(train_df):
    tr, va = train_test_split(
        train_df, test_size=VAL_FRAC_OF_TRAIN, stratify=train_df["y"], random_state=SEED
    )
    return tr.reset_index(drop=True), va.reset_index(drop=True)


train_df, test_df = train_test_split(df, test_size=TEST_SIZE, stratify=df["y"], random_state=SEED)
train_df, test_df = train_df.reset_index(drop=True), test_df.reset_index(drop=True)

if USE_TEST_AS_VAL:
    val_df = test_df
else:
    train_df, val_df = split_train_val(train_df)

print(f"train={len(train_df)}  val={len(val_df)}  test={len(test_df)}")

# %% 4. tf.data input pipeline with real-time augmentation
AUTOTUNE = tf.data.AUTOTUNE

# IMPROVEMENT 5: richer augmentation (adds translation and contrast on top of the original four)
augment = keras.Sequential(
    [
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(20 / 360),                       # about +/-20 degrees
        layers.RandomZoom(0.2),
        layers.RandomTranslation(0.1, 0.1),
        layers.RandomBrightness(0.2, value_range=(0, 255)),
        layers.RandomContrast(0.2),
    ],
    name="augment",
)


def _load(path, y):
    img = tf.io.read_file(path)
    img = tf.io.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = tf.cast(tf.round(img), tf.uint8)          # uint8 so the cache stays small
    return img, tf.one_hot(y, NUM_CLASSES)


def make_ds(frame, training):
    ds = tf.data.Dataset.from_tensor_slices((frame["path"].values, frame["y"].values))
    ds = ds.map(_load, num_parallel_calls=AUTOTUNE).cache()   # decode + resize only once
    if training:
        ds = ds.shuffle(1000, seed=SEED, reshuffle_each_iteration=True)
        ds = ds.map(
            lambda x, y: (augment(tf.cast(x, tf.float32), training=True), y),
            num_parallel_calls=AUTOTUNE,
        )
    ds = ds.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y), num_parallel_calls=AUTOTUNE)
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)


# %% 5. Model: DenseNet121 + custom head, with block-wise freezing
# Keras DenseNet121 layer naming:
#   conv1_*            stem
#   conv2_block*       dense block 1 (6 layers)     pool2_*  transition 1
#   conv3_block*       dense block 2 (12 layers)    pool3_*  transition 2
#   conv4_block*       dense block 3 (24 layers)    pool4_*  transition 3
#   conv5_block*       dense block 4 (16 layers)
#   bn                 final batch norm
# IMPROVEMENT 6: Phase 1 trains ONLY the new head (random head + pretrained block 4 at lr 1e-3
# damages the pretrained weights). Phase 2 then fine-tunes the entire backbone at a low lr.
PHASE1_TRAINABLE = ()          # head only
PHASE2_TRAINABLE = None        # None = whole backbone trainable


def build_model():
    base = DenseNet121(include_top=False, weights="imagenet", input_shape=(*IMG_SIZE, 3))
    x = layers.GlobalAveragePooling2D(name="gap")(base.output)
    x = layers.Dropout(DROPOUT_1, name="drop1")(x)
    x = layers.Dense(DENSE_UNITS, activation="relu", kernel_initializer="glorot_uniform", name="fc256")(x)
    x = layers.Dropout(DROPOUT_2, name="drop2")(x)
    # float32 output layer so softmax is numerically safe under mixed precision
    out = layers.Dense(NUM_CLASSES, activation="softmax", kernel_initializer="glorot_uniform",
                       dtype="float32", name="out")(x)
    model = keras.Model(base.input, out, name="densenet121_rice")
    base_names = {l.name for l in base.layers}
    return model, base_names


def set_trainable(model, base_names, prefixes):
    for l in model.layers:
        if l.name in base_names:
            if prefixes is None:
                l.trainable = True
            elif len(prefixes) == 0:
                l.trainable = False
            else:
                l.trainable = l.name.startswith(tuple(prefixes))
        else:
            l.trainable = True  # classification head


def compile_model(model, lr):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss=keras.losses.CategoricalCrossentropy(label_smoothing=LABEL_SMOOTHING),
        metrics=["accuracy"],
    )


def make_callbacks():
    return [
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=LR_FACTOR, patience=LR_PATIENCE, min_lr=1e-7, verbose=1
        ),
        keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=ES_PATIENCE, restore_best_weights=True, verbose=1
        ),
    ]


def merge_history(h1, h2):
    keys = [k for k in h1.history if k in h2.history]
    return {k: h1.history[k] + h2.history[k] for k in keys}


def get_class_weights(frame):
    if not USE_CLASS_WEIGHTS:
        return None
    w = compute_class_weight("balanced", classes=np.arange(NUM_CLASSES), y=frame["y"].values)
    w = np.sqrt(w)                                   # softened so rare classes are not over-pushed
    return {i: float(v) for i, v in enumerate(w)}


def train_two_phase(train_ds, val_ds, class_weight=None, verbose=1):
    model, base_names = build_model()

    # Phase 1 - backbone frozen, train head only, lr = 1e-3
    set_trainable(model, base_names, PHASE1_TRAINABLE)
    compile_model(model, LR_PHASE1)
    h1 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE1_EPOCHS,
                   callbacks=make_callbacks(), class_weight=class_weight, verbose=verbose)

    # Phase 2 - unfreeze the whole backbone, lr = 1e-4
    set_trainable(model, base_names, PHASE2_TRAINABLE)
    compile_model(model, LR_PHASE2)
    h2 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE2_EPOCHS,
                   callbacks=make_callbacks(), class_weight=class_weight, verbose=verbose)

    return model, merge_history(h1, h2)


# %% 6. Metrics (Eq. 1-4 of the paper, one-vs-rest per class) -- UNCHANGED
def per_class_metrics(cm):
    cm = cm.astype(np.float64)
    tp = np.diag(cm)
    fn = cm.sum(axis=1) - tp
    fp = cm.sum(axis=0) - tp
    tn = cm.sum() - tp - fn - fp
    eps = 1e-12
    sens = tp / (tp + fn + eps)                        # Eq. 1 (recall)
    spec = tn / (tn + fp + eps)                        # Eq. 2
    acc = (tp + tn) / (tp + tn + fp + fn + eps)        # Eq. 3 (per-class, one-vs-rest)
    prec = tp / (tp + fp + eps)                        # Eq. 4
    f1 = 2 * prec * sens / (prec + sens + eps)
    return sens, spec, prec, f1, acc


def predict_probs(model, ds):
    """Softmax probabilities; with USE_TTA averages original + h-flip + v-flip + both flips."""
    p = model.predict(ds, verbose=0)
    if not USE_TTA:
        return p

    def flipped(fn):
        return ds.map(lambda x, y: (fn(x), y), num_parallel_calls=AUTOTUNE)

    p += model.predict(flipped(lambda x: tf.image.flip_left_right(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(tf.image.flip_left_right(x))), verbose=0)
    return p / 4.0


def evaluate(model, frame):
    ds = make_ds(frame, training=False)          # not shuffled, order matches frame
    y_true = frame["y"].values
    y_pred = predict_probs(model, ds).argmax(axis=1)
    cm = confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES)))

    sens, spec, prec, f1, acc = per_class_metrics(cm)
    table4 = pd.DataFrame(
        {
            "Sensitivity": sens * 100,
            "Specificity": spec * 100,
            "Precision": prec * 100,
            "F-score": f1 * 100,
            "Accuracy": acc * 100,
        },
        index=PRETTY,
    ).round(2)
    table4.index.name = "Disease Types"

    overall_acc = np.trace(cm) / cm.sum() * 100
    table5 = pd.DataFrame(
        {
            "Sensitivity": [sens.mean() * 100],
            "Specificity": [spec.mean() * 100],
            "Precision": [prec.mean() * 100],
            "F-score": [f1.mean() * 100],
            "Accuracy": [overall_acc],
        },
        index=["Our proposed model"],
    ).round(2)
    return cm, table4, table5, y_true, y_pred


# %% 7. Plots (Fig. 4 and Fig. 5) -- saved to OUT_DIR, then displayed
def plot_curves(hist, path):
    ep = np.arange(1, len(hist["loss"]) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
    ax[0].plot(ep, hist["loss"], "o-", ms=3, label="Training Loss")
    ax[0].plot(ep, hist["val_loss"], "x-", ms=4, label="Validation Loss")
    ax[0].set(title="Training and Validation Loss", xlabel="Epochs", ylabel="Loss")
    ax[1].plot(ep, hist["accuracy"], "o-", ms=3, label="Training Accuracy")
    ax[1].plot(ep, hist["val_accuracy"], "x-", ms=4, label="Validation Accuracy")
    ax[1].set(title="Training and Validation Accuracy", xlabel="Epochs", ylabel="Accuracy")
    for a in ax:
        a.grid(alpha=0.3)
        a.legend()
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    print("Saved:", path)
    plt.show()
    plt.close(fig)


def plot_confusion(cm, path):
    cmn = cm / cm.sum(axis=1, keepdims=True)
    fig = plt.figure(figsize=(9, 7))
    sns.heatmap(
        cmn, annot=True, fmt=".2%", cmap="Blues", vmin=0, vmax=1,
        xticklabels=PRETTY, yticklabels=PRETTY, cbar_kws={"label": "Percentage"},
    )
    plt.title("Confusion Matrix for Rice Disease Classification (Normalized by Row)")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    print("Saved:", path)
    plt.show()
    plt.close(fig)


# %% 8. Main experiment: train once, evaluate on the held-out test split
if RUN_MAIN:
    train_ds = make_ds(train_df, training=True)
    val_ds = make_ds(val_df, training=False)

    model, history = train_two_phase(train_ds, val_ds, class_weight=get_class_weights(train_df))
    model.save(OUT_DIR / "densenet121_rice.keras")
    pd.DataFrame(history).to_csv(OUT_DIR / "history.csv", index=False)

    cm, table4, table5, y_true, y_pred = evaluate(model, test_df)

    print("\nTable 4 - per-class results (%)")
    print(table4)
    print("\nTable 5 - overall results (%)")
    print(table5)

    table4.to_csv(OUT_DIR / "table4_per_class.csv")
    table5.to_csv(OUT_DIR / "table5_overall.csv")
    np.save(OUT_DIR / "confusion_matrix.npy", cm)

    plot_curves(history, OUT_DIR / "fig4_curves.png")
    plot_confusion(cm, OUT_DIR / "fig5_confusion_matrix.png")


# %% 9. 5-fold cross-validation (Table 7) and 95% CI / significance (Table 8) -- UNCHANGED logic
def run_cross_validation():
    skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
    rows = []
    for k, (tr_idx, te_idx) in enumerate(skf.split(df, df["y"]), start=1):
        print(f"\n===== Fold {k}/{N_FOLDS} =====")
        keras.backend.clear_session()
        tr_full = df.iloc[tr_idx].reset_index(drop=True)
        te = df.iloc[te_idx].reset_index(drop=True)
        tr, va = split_train_val(tr_full)

        m, _ = train_two_phase(make_ds(tr, True), make_ds(va, False),
                               class_weight=get_class_weights(tr), verbose=2)

        y_pred = predict_probs(m, make_ds(te, False)).argmax(axis=1)
        y_true = te["y"].values
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)
        rows.append(
            {
                "Fold": f"Fold {k}",
                "Accuracy (%)": accuracy_score(y_true, y_pred) * 100,
                "Precision (%)": p * 100,
                "Recall (%)": r * 100,
                "F1-Score (%)": f * 100,
            }
        )
        print(rows[-1])
        del m

    cv = pd.DataFrame(rows).set_index("Fold")
    metrics = list(cv.columns)

    # Table 7: mean +/- SD (sample SD, ddof=1)
    mean, sd = cv.mean(), cv.std(ddof=1)
    table7 = cv.round(2).copy()
    table7.loc["Mean ± SD"] = [f"{mean[c]:.1f} ± {sd[c]:.2f}" for c in metrics]
    print("\nTable 7 - 5-fold cross-validation")
    print(table7)

    # Table 8: t-based 95% CI over folds + one-sample t-test against a reference baseline
    n = len(cv)
    tcrit = stats.t.ppf(0.975, df=n - 1)
    t8 = []
    for c in metrics:
        half = tcrit * sd[c] / np.sqrt(n)
        _, p = stats.ttest_1samp(cv[c].values, BASELINE_PCT)
        t8.append(
            {
                "Metric": c.replace(" (%)", ""),
                "Mean (%)": round(mean[c], 1),
                "95% CI": f"[{mean[c] - half:.1f}, {mean[c] + half:.1f}]",
                "p-value": "<0.001" if p < 0.001 else f"{p:.3f}",
            }
        )
    table8 = pd.DataFrame(t8).set_index("Metric")
    print(f"\nTable 8 - 95% CI and significance (vs. {BASELINE_PCT}% baseline)")
    print(table8)

    table7.to_csv(OUT_DIR / "table7_cv.csv")
    table8.to_csv(OUT_DIR / "table8_ci.csv")
    return cv, table7, table8


if RUN_CV:
    cv_raw, table7, table8 = run_cross_validation()
```

```python
import os
import random
from pathlib import Path

import matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.applications.densenet import preprocess_input
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, precision_recall_fscore_support, accuracy_score
from sklearn.utils.class_weight import compute_class_weight
from scipy import stats

# ---- reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
keras.utils.set_random_seed(SEED)

# ---- data
# The paper uses 7 disease classes. Kaggle's Paddy set has 10 (adds dead_heart, hispa, normal).
# Set KEEP_CLASSES = None to use all 10.
KEEP_CLASSES = [
    "bacterial_leaf_blight",
    "bacterial_leaf_streak",
    "bacterial_panicle_blight",
    "blast",
    "brown_spot",
    "downy_mildew",
    "tungro",
]
# IMPROVEMENT 1: higher resolution, keeping the 3:4 aspect ratio of the original photos
# (originals are 480x640; squashing to 224x224 destroys small lesion detail).
# DenseNet121 + GlobalAveragePooling accepts any size. If you hit out-of-memory, use (256, 320).
IMG_SIZE = (320, 448)          # (height, width). IMPROVEMENT 7: more lesion detail than 288x384.
                               # Out of memory? -> (288, 384), or lower BATCH_SIZE.
TEST_SIZE = 0.20
VAL_FRAC_OF_TRAIN = 0.10       # validation carved out of the training split (used by callbacks)
USE_TEST_AS_VAL = False        # keep False: using the test set for callbacks leaks information

# ---- training
BATCH_SIZE = 16                # smaller batches at the larger resolution; lower to 8 if out of memory
N_ENSEMBLE = 3                 # IMPROVEMENT 9: models averaged at test time (1 = single model, fastest)
PHASE1_EPOCHS = 8              # warm-up: new head only (backbone frozen)
PHASE2_EPOCHS = 60            # fine-tune the whole backbone (early stopping decides the real length)
LR_PHASE1 = 1e-3
LR_PHASE2 = 5e-5               # IMPROVEMENT 8: gentler fine-tuning (smaller batch + bigger images)
DROPOUT_1, DROPOUT_2, DENSE_UNITS = 0.5, 0.3, 256
LR_FACTOR, LR_PATIENCE = 0.3, 2
ES_PATIENCE = 10
LABEL_SMOOTHING = 0.1          # IMPROVEMENT 2: reduces over-confidence, usually +0.5-1% on noisy field images
USE_CLASS_WEIGHTS = True       # IMPROVEMENT 3: dataset is imbalanced (blast ~1700 vs panicle blight ~340)
USE_TTA = True                 # IMPROVEMENT 4: flip test-time augmentation at evaluation
USE_MIXED_PRECISION = True     # faster and lighter on memory (needs a GPU)

# ---- experiments
RUN_MAIN = True
RUN_CV = False                 # 5 full trainings, expensive. Turn on when the main run looks right.
N_FOLDS = 5
BASELINE_PCT = 94.0            # Saputra et al. DenseNet121 (Table 6), reference for the t-test

if USE_MIXED_PRECISION and tf.config.list_physical_devices("GPU"):
    keras.mixed_precision.set_global_policy("mixed_float16")

# ---- paths
OUT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)


def find_data_root():
    candidates = [
        "/kaggle/input/paddy-disease-classification",
        "/content/paddy-disease-classification",
        "./paddy-disease-classification",
    ]
    for c in candidates:
        if (Path(c) / "train.csv").exists():
            return Path(c)
    # Colab fallback: needs ~/.kaggle/kaggle.json and the competition rules accepted on the website
    os.system("pip -q install kaggle")
    os.system("kaggle competitions download -c paddy-disease-classification -p /content")
    os.system("unzip -q -o /content/paddy-disease-classification.zip -d /content/paddy-disease-classification")
    return Path("/content/paddy-disease-classification")


DATA_ROOT = find_data_root()
print("Data root:", DATA_ROOT)
print("GPUs:", tf.config.list_physical_devices("GPU"))

# %% 2. Load metadata and build the dataframe
# train.csv columns: image_id, label, variety, age. Images: train_images/<label>/<image_id>
# (test_images has no labels, so everything is split from train_images.)
df = pd.read_csv(DATA_ROOT / "train.csv")
if KEEP_CLASSES is not None:
    df = df[df["label"].isin(KEEP_CLASSES)].reset_index(drop=True)

CLASSES = sorted(df["label"].unique())
NUM_CLASSES = len(CLASSES)
label2idx = {c: i for i, c in enumerate(CLASSES)}
df["y"] = df["label"].map(label2idx)
df["path"] = [str(DATA_ROOT / "train_images" / l / i) for i, l in zip(df["image_id"], df["label"])]
PRETTY = [c.replace("_", " ").capitalize() for c in CLASSES]

print(f"{len(df)} images, {NUM_CLASSES} classes")
print(df["label"].value_counts().rename("n_images"))


# %% 3. Splits
def split_train_val(train_df):
    tr, va = train_test_split(
        train_df, test_size=VAL_FRAC_OF_TRAIN, stratify=train_df["y"], random_state=SEED
    )
    return tr.reset_index(drop=True), va.reset_index(drop=True)


train_df, test_df = train_test_split(df, test_size=TEST_SIZE, stratify=df["y"], random_state=SEED)
train_df, test_df = train_df.reset_index(drop=True), test_df.reset_index(drop=True)

if USE_TEST_AS_VAL:
    val_df = test_df
else:
    train_df, val_df = split_train_val(train_df)

print(f"train={len(train_df)}  val={len(val_df)}  test={len(test_df)}")

# %% 4. tf.data input pipeline with real-time augmentation
AUTOTUNE = tf.data.AUTOTUNE

# IMPROVEMENT 5: richer augmentation (adds translation and contrast on top of the original four)
augment = keras.Sequential(
    [
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(20 / 360),                       # about +/-20 degrees
        layers.RandomZoom(0.2),
        layers.RandomTranslation(0.1, 0.1),
        layers.RandomBrightness(0.2, value_range=(0, 255)),
        layers.RandomContrast(0.2),
    ],
    name="augment",
)


def _load(path, y):
    img = tf.io.read_file(path)
    img = tf.io.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = tf.cast(tf.round(img), tf.uint8)          # uint8 so the cache stays small
    return img, tf.one_hot(y, NUM_CLASSES)


def make_ds(frame, training):
    ds = tf.data.Dataset.from_tensor_slices((frame["path"].values, frame["y"].values))
    ds = ds.map(_load, num_parallel_calls=AUTOTUNE).cache()   # decode + resize only once
    if training:
        ds = ds.shuffle(1000, seed=SEED, reshuffle_each_iteration=True)
        ds = ds.map(
            lambda x, y: (augment(tf.cast(x, tf.float32), training=True), y),
            num_parallel_calls=AUTOTUNE,
        )
    ds = ds.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y), num_parallel_calls=AUTOTUNE)
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)


# %% 5. Model: DenseNet121 + custom head, with block-wise freezing
# Keras DenseNet121 layer naming:
#   conv1_*            stem
#   conv2_block*       dense block 1 (6 layers)     pool2_*  transition 1
#   conv3_block*       dense block 2 (12 layers)    pool3_*  transition 2
#   conv4_block*       dense block 3 (24 layers)    pool4_*  transition 3
#   conv5_block*       dense block 4 (16 layers)
#   bn                 final batch norm
# IMPROVEMENT 6: Phase 1 trains ONLY the new head (random head + pretrained block 4 at lr 1e-3
# damages the pretrained weights). Phase 2 then fine-tunes the entire backbone at a low lr.
PHASE1_TRAINABLE = ()          # head only
PHASE2_TRAINABLE = None        # None = whole backbone trainable


def build_model():
    base = DenseNet121(include_top=False, weights="imagenet", input_shape=(*IMG_SIZE, 3))
    x = layers.GlobalAveragePooling2D(name="gap")(base.output)
    x = layers.Dropout(DROPOUT_1, name="drop1")(x)
    x = layers.Dense(DENSE_UNITS, activation="relu", kernel_initializer="glorot_uniform", name="fc256")(x)
    x = layers.Dropout(DROPOUT_2, name="drop2")(x)
    # float32 output layer so softmax is numerically safe under mixed precision
    out = layers.Dense(NUM_CLASSES, activation="softmax", kernel_initializer="glorot_uniform",
                       dtype="float32", name="out")(x)
    model = keras.Model(base.input, out, name="densenet121_rice")
    base_names = {l.name for l in base.layers}
    return model, base_names


def set_trainable(model, base_names, prefixes):
    for l in model.layers:
        if l.name in base_names:
            if prefixes is None:
                l.trainable = True
            elif len(prefixes) == 0:
                l.trainable = False
            else:
                l.trainable = l.name.startswith(tuple(prefixes))
        else:
            l.trainable = True  # classification head


def compile_model(model, lr):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss=keras.losses.CategoricalCrossentropy(label_smoothing=LABEL_SMOOTHING),
        metrics=["accuracy"],
    )


def make_callbacks():
    return [
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=LR_FACTOR, patience=LR_PATIENCE, min_lr=1e-7, verbose=1
        ),
        keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=ES_PATIENCE, restore_best_weights=True, verbose=1
        ),
    ]


def merge_history(h1, h2):
    keys = [k for k in h1.history if k in h2.history]
    return {k: h1.history[k] + h2.history[k] for k in keys}


def get_class_weights(frame):
    if not USE_CLASS_WEIGHTS:
        return None
    w = compute_class_weight("balanced", classes=np.arange(NUM_CLASSES), y=frame["y"].values)
    w = np.sqrt(w)                                   # softened so rare classes are not over-pushed
    return {i: float(v) for i, v in enumerate(w)}


def train_two_phase(train_ds, val_ds, class_weight=None, verbose=1):
    model, base_names = build_model()

    # Phase 1 - backbone frozen, train head only, lr = 1e-3
    set_trainable(model, base_names, PHASE1_TRAINABLE)
    compile_model(model, LR_PHASE1)
    h1 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE1_EPOCHS,
                   callbacks=make_callbacks(), class_weight=class_weight, verbose=verbose)

    # Phase 2 - unfreeze the whole backbone, lr = 1e-4
    set_trainable(model, base_names, PHASE2_TRAINABLE)
    compile_model(model, LR_PHASE2)
    h2 = model.fit(train_ds, validation_data=val_ds, epochs=PHASE2_EPOCHS,
                   callbacks=make_callbacks(), class_weight=class_weight, verbose=verbose)

    return model, merge_history(h1, h2)


# %% 6. Metrics (Eq. 1-4 of the paper, one-vs-rest per class) -- UNCHANGED
def per_class_metrics(cm):
    cm = cm.astype(np.float64)
    tp = np.diag(cm)
    fn = cm.sum(axis=1) - tp
    fp = cm.sum(axis=0) - tp
    tn = cm.sum() - tp - fn - fp
    eps = 1e-12
    sens = tp / (tp + fn + eps)                        # Eq. 1 (recall)
    spec = tn / (tn + fp + eps)                        # Eq. 2
    acc = (tp + tn) / (tp + tn + fp + fn + eps)        # Eq. 3 (per-class, one-vs-rest)
    prec = tp / (tp + fp + eps)                        # Eq. 4
    f1 = 2 * prec * sens / (prec + sens + eps)
    return sens, spec, prec, f1, acc


def predict_probs(model, ds):
    """Softmax probabilities; with USE_TTA averages original + h-flip + v-flip + both flips."""
    p = model.predict(ds, verbose=0)
    if not USE_TTA:
        return p

    def flipped(fn):
        return ds.map(lambda x, y: (fn(x), y), num_parallel_calls=AUTOTUNE)

    p += model.predict(flipped(lambda x: tf.image.flip_left_right(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(tf.image.flip_left_right(x))), verbose=0)
    return p / 4.0


def evaluate_probs(probs, frame):
    """Same metrics as before, computed from (ensembled) softmax probabilities."""
    y_true = frame["y"].values
    y_pred = probs.argmax(axis=1)
    cm = confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES)))

    sens, spec, prec, f1, acc = per_class_metrics(cm)
    table4 = pd.DataFrame(
        {
            "Sensitivity": sens * 100,
            "Specificity": spec * 100,
            "Precision": prec * 100,
            "F-score": f1 * 100,
            "Accuracy": acc * 100,
        },
        index=PRETTY,
    ).round(2)
    table4.index.name = "Disease Types"

    overall_acc = np.trace(cm) / cm.sum() * 100
    table5 = pd.DataFrame(
        {
            "Sensitivity": [sens.mean() * 100],
            "Specificity": [spec.mean() * 100],
            "Precision": [prec.mean() * 100],
            "F-score": [f1.mean() * 100],
            "Accuracy": [overall_acc],
        },
        index=["Our proposed model"],
    ).round(2)
    return cm, table4, table5, y_true, y_pred


# %% 7. Plots (Fig. 4 and Fig. 5) -- saved to OUT_DIR, then displayed
def plot_curves(hist, path):
    ep = np.arange(1, len(hist["loss"]) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
    ax[0].plot(ep, hist["loss"], "o-", ms=3, label="Training Loss")
    ax[0].plot(ep, hist["val_loss"], "x-", ms=4, label="Validation Loss")
    ax[0].set(title="Training and Validation Loss", xlabel="Epochs", ylabel="Loss")
    ax[1].plot(ep, hist["accuracy"], "o-", ms=3, label="Training Accuracy")
    ax[1].plot(ep, hist["val_accuracy"], "x-", ms=4, label="Validation Accuracy")
    ax[1].set(title="Training and Validation Accuracy", xlabel="Epochs", ylabel="Accuracy")
    for a in ax:
        a.grid(alpha=0.3)
        a.legend()
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    print("Saved:", path)
    plt.show()
    plt.close(fig)


def plot_confusion(cm, path):
    cmn = cm / cm.sum(axis=1, keepdims=True)
    fig = plt.figure(figsize=(9, 7))
    sns.heatmap(
        cmn, annot=True, fmt=".2%", cmap="Blues", vmin=0, vmax=1,
        xticklabels=PRETTY, yticklabels=PRETTY, cbar_kws={"label": "Percentage"},
    )
    plt.title("Confusion Matrix for Rice Disease Classification (Normalized by Row)")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(path, dpi=300, bbox_inches="tight")
    print("Saved:", path)
    plt.show()
    plt.close(fig)


# %% 8. Main experiment: train once, evaluate on the held-out test split
if RUN_MAIN:
    train_ds = make_ds(train_df, training=True)
    val_ds = make_ds(val_df, training=False)

    # IMPROVEMENT 9: seed ensemble. N_ENSEMBLE models are trained on the SAME split with different
    # seeds (different head init, augmentation and batch order); their softmax outputs are averaged.
    # Each model's test probabilities are saved to disk, so an interrupted run can be resumed.
    test_ds = make_ds(test_df, training=False)
    cw = get_class_weights(train_df)
    all_probs, history = [], None
    for k in range(N_ENSEMBLE):
        prob_file = OUT_DIR / f"test_probs_model{k + 1}.npy"
        if prob_file.exists():
            print(f"\n[model {k + 1}/{N_ENSEMBLE}] found {prob_file.name}, skipping training")
            all_probs.append(np.load(prob_file))
            continue
        print(f"\n===== Training ensemble member {k + 1}/{N_ENSEMBLE} =====")
        keras.backend.clear_session()
        keras.utils.set_random_seed(SEED + k)
        model, hist_k = train_two_phase(train_ds, val_ds, class_weight=cw)
        model.save(OUT_DIR / f"densenet121_rice_model{k + 1}.keras")
        pd.DataFrame(hist_k).to_csv(OUT_DIR / f"history_model{k + 1}.csv", index=False)
        plot_curves(hist_k, OUT_DIR / f"fig4_curves_model{k + 1}.png")
        p_k = predict_probs(model, test_ds)
        np.save(prob_file, p_k)
        single_acc = (p_k.argmax(1) == test_df["y"].values).mean() * 100
        print(f"model {k + 1} alone: test accuracy {single_acc:.2f}%")
        all_probs.append(p_k)
        if history is None:
            history = hist_k
        del model

    ens_probs = np.mean(all_probs, axis=0)
    cm, table4, table5, y_true, y_pred = evaluate_probs(ens_probs, test_df)
    print(f"\nEnsemble of {len(all_probs)} model(s): test accuracy {table5['Accuracy'].iloc[0]:.2f}%")

    print("\nTable 4 - per-class results (%)")
    print(table4)
    print("\nTable 5 - overall results (%)")
    print(table5)

    table4.to_csv(OUT_DIR / "table4_per_class.csv")
    table5.to_csv(OUT_DIR / "table5_overall.csv")
    np.save(OUT_DIR / "confusion_matrix.npy", cm)

    if history is not None:                      # None only if every model was loaded from disk
        pd.DataFrame(history).to_csv(OUT_DIR / "history.csv", index=False)
        plot_curves(history, OUT_DIR / "fig4_curves.png")
    plot_confusion(cm, OUT_DIR / "fig5_confusion_matrix.png")


# %% 9. 5-fold cross-validation (Table 7) and 95% CI / significance (Table 8) -- UNCHANGED logic
def run_cross_validation():
    skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
    rows = []
    for k, (tr_idx, te_idx) in enumerate(skf.split(df, df["y"]), start=1):
        print(f"\n===== Fold {k}/{N_FOLDS} =====")
        keras.backend.clear_session()
        tr_full = df.iloc[tr_idx].reset_index(drop=True)
        te = df.iloc[te_idx].reset_index(drop=True)
        tr, va = split_train_val(tr_full)

        m, _ = train_two_phase(make_ds(tr, True), make_ds(va, False),
                               class_weight=get_class_weights(tr), verbose=2)

        y_pred = predict_probs(m, make_ds(te, False)).argmax(axis=1)
        y_true = te["y"].values
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)
        rows.append(
            {
                "Fold": f"Fold {k}",
                "Accuracy (%)": accuracy_score(y_true, y_pred) * 100,
                "Precision (%)": p * 100,
                "Recall (%)": r * 100,
                "F1-Score (%)": f * 100,
            }
        )
        print(rows[-1])
        del m

    cv = pd.DataFrame(rows).set_index("Fold")
    metrics = list(cv.columns)

    # Table 7: mean +/- SD (sample SD, ddof=1)
    mean, sd = cv.mean(), cv.std(ddof=1)
    table7 = cv.round(2).copy()
    table7.loc["Mean ± SD"] = [f"{mean[c]:.1f} ± {sd[c]:.2f}" for c in metrics]
    print("\nTable 7 - 5-fold cross-validation")
    print(table7)

    # Table 8: t-based 95% CI over folds + one-sample t-test against a reference baseline
    n = len(cv)
    tcrit = stats.t.ppf(0.975, df=n - 1)
    t8 = []
    for c in metrics:
        half = tcrit * sd[c] / np.sqrt(n)
        _, p = stats.ttest_1samp(cv[c].values, BASELINE_PCT)
        t8.append(
            {
                "Metric": c.replace(" (%)", ""),
                "Mean (%)": round(mean[c], 1),
                "95% CI": f"[{mean[c] - half:.1f}, {mean[c] + half:.1f}]",
                "p-value": "<0.001" if p < 0.001 else f"{p:.3f}",
            }
        )
    table8 = pd.DataFrame(t8).set_index("Metric")
    print(f"\nTable 8 - 95% CI and significance (vs. {BASELINE_PCT}% baseline)")
    print(table8)

    table7.to_csv(OUT_DIR / "table7_cv.csv")
    table8.to_csv(OUT_DIR / "table8_ci.csv")
    return cv, table7, table8


if RUN_CV:
    cv_raw, table7, table8 = run_cross_validation()
```

```python
import os
import gc
import json
import random
from pathlib import Path

import matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import DenseNet121
from tensorflow.keras.applications.densenet import preprocess_input
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, precision_recall_fscore_support, accuracy_score
from sklearn.utils.class_weight import compute_class_weight
from scipy import stats

# ---- reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
keras.utils.set_random_seed(SEED)

# ---- data
# The paper uses 7 disease classes. Kaggle's Paddy set has 10 (adds dead_heart, hispa, normal).
# Set KEEP_CLASSES = None to use all 10.
KEEP_CLASSES = [
    "bacterial_leaf_blight",
    "bacterial_leaf_streak",
    "bacterial_panicle_blight",
    "blast",
    "brown_spot",
    "downy_mildew",
    "tungro",
]
# IMPROVEMENT 1: higher resolution, keeping the 3:4 aspect ratio of the original photos
# (originals are 480x640; squashing to 224x224 destroys small lesion detail).
# DenseNet121 + GlobalAveragePooling accepts any size. If you hit out-of-memory, use (256, 320).
IMG_SIZE = (288, 384)          # (height, width)
TEST_SIZE = 0.20
VAL_FRAC_OF_TRAIN = 0.10       # validation carved out of the training split (used by callbacks)
USE_TEST_AS_VAL = False        # keep False: using the test set for callbacks leaks information

# ---- training
BATCH_SIZE = 32                # lower to 16 if the GPU runs out of memory
PHASE1_EPOCHS = 10             # warm-up: new head only (backbone frozen)            (was 8)
PHASE2_EPOCHS = 80             # fine-tune whole backbone; early stopping ends it    (was 45)
LR_FACTOR, LR_PATIENCE = 0.3, 3          # LR x0.3 after 3 epochs without val_loss improvement (was 2)
ES_PATIENCE = 12               # was 8: gives the LR schedule room to drop several times before stopping

# Default hyper-parameters = exactly the settings that gave you 96.7%.
# If RUN_TUNING is True, the best configuration found on the VALIDATION set replaces these.
HP_DEFAULT = dict(lr1=1e-3, lr2=1e-4, drop1=0.5, drop2=0.3, dense=256, ls=0.1)

# ---- hyper-parameter search (scored on the validation split only, never on the test split)
RUN_TUNING = True
N_TRIALS = 8                   # trial 0 is always HP_DEFAULT, so tuning can never end up worse on validation
TUNE_P1_EPOCHS, TUNE_P2_EPOCHS, TUNE_ES_PATIENCE = 3, 15, 4   # short runs, only to rank configurations
SEARCH_SPACE = dict(
    lr1=[5e-4, 1e-3],
    lr2=[3e-5, 5e-5, 1e-4, 2e-4],
    drop1=[0.3, 0.5, 0.6],
    drop2=[0.2, 0.3, 0.4],
    dense=[256, 512],
    ls=[0.0, 0.05, 0.1, 0.15],
)
N_ENSEMBLE = 1                 # >1 trains several seeds and averages them (most reliable extra gain, N x time)
RESUME = False                 # True = reuse test_probs_modelK.npy from OUT_DIR (delete them if settings changed)
USE_CLASS_WEIGHTS = True       # IMPROVEMENT 3: dataset is imbalanced (blast ~1700 vs panicle blight ~340)
USE_TTA = True                 # IMPROVEMENT 4: flip test-time augmentation at evaluation
USE_MIXED_PRECISION = True     # faster and lighter on memory (needs a GPU)

# ---- experiments
RUN_MAIN = True
RUN_CV = False                 # 5 full trainings, expensive. Turn on when the main run looks right.
N_FOLDS = 5
BASELINE_PCT = 94.0            # Saputra et al. DenseNet121 (Table 6), reference for the t-test

if USE_MIXED_PRECISION and tf.config.list_physical_devices("GPU"):
    keras.mixed_precision.set_global_policy("mixed_float16")

# ---- paths
OUT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)


def find_data_root():
    candidates = [
        "/kaggle/input/paddy-disease-classification",
        "/content/paddy-disease-classification",
        "./paddy-disease-classification",
    ]
    for c in candidates:
        if (Path(c) / "train.csv").exists():
            return Path(c)
    # Colab fallback: needs ~/.kaggle/kaggle.json and the competition rules accepted on the website
    os.system("pip -q install kaggle")
    os.system("kaggle competitions download -c paddy-disease-classification -p /content")
    os.system("unzip -q -o /content/paddy-disease-classification.zip -d /content/paddy-disease-classification")
    return Path("/content/paddy-disease-classification")


DATA_ROOT = find_data_root()
print("Data root:", DATA_ROOT)
print("GPUs:", tf.config.list_physical_devices("GPU"))

# %% 2. Load metadata and build the dataframe
# train.csv columns: image_id, label, variety, age. Images: train_images/<label>/<image_id>
# (test_images has no labels, so everything is split from train_images.)
df = pd.read_csv(DATA_ROOT / "train.csv")
if KEEP_CLASSES is not None:
    df = df[df["label"].isin(KEEP_CLASSES)].reset_index(drop=True)

CLASSES = sorted(df["label"].unique())
NUM_CLASSES = len(CLASSES)
label2idx = {c: i for i, c in enumerate(CLASSES)}
df["y"] = df["label"].map(label2idx)
df["path"] = [str(DATA_ROOT / "train_images" / l / i) for i, l in zip(df["image_id"], df["label"])]
PRETTY = [c.replace("_", " ").capitalize() for c in CLASSES]

print(f"{len(df)} images, {NUM_CLASSES} classes")
print(df["label"].value_counts().rename("n_images"))


# %% 3. Splits
def split_train_val(train_df):
    tr, va = train_test_split(
        train_df, test_size=VAL_FRAC_OF_TRAIN, stratify=train_df["y"], random_state=SEED
    )
    return tr.reset_index(drop=True), va.reset_index(drop=True)


train_df, test_df = train_test_split(df, test_size=TEST_SIZE, stratify=df["y"], random_state=SEED)
train_df, test_df = train_df.reset_index(drop=True), test_df.reset_index(drop=True)

if USE_TEST_AS_VAL:
    val_df = test_df
else:
    train_df, val_df = split_train_val(train_df)

print(f"train={len(train_df)}  val={len(val_df)}  test={len(test_df)}")

# %% 4. tf.data input pipeline with real-time augmentation
AUTOTUNE = tf.data.AUTOTUNE

# IMPROVEMENT 5: richer augmentation (adds translation and contrast on top of the original four)
augment = keras.Sequential(
    [
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(20 / 360),                       # about +/-20 degrees
        layers.RandomZoom(0.2),
        layers.RandomTranslation(0.1, 0.1),
        layers.RandomBrightness(0.2, value_range=(0, 255)),
        layers.RandomContrast(0.2),
    ],
    name="augment",
)


def _load(path, y):
    img = tf.io.read_file(path)
    img = tf.io.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, IMG_SIZE)
    img = tf.cast(tf.round(img), tf.uint8)          # uint8 so the cache stays small
    return img, tf.one_hot(y, NUM_CLASSES)


def make_ds(frame, training):
    ds = tf.data.Dataset.from_tensor_slices((frame["path"].values, frame["y"].values))
    ds = ds.map(_load, num_parallel_calls=AUTOTUNE).cache()   # decode + resize only once
    if training:
        ds = ds.shuffle(1000, seed=SEED, reshuffle_each_iteration=True)
        ds = ds.map(
            lambda x, y: (augment(tf.cast(x, tf.float32), training=True), y),
            num_parallel_calls=AUTOTUNE,
        )
    ds = ds.map(lambda x, y: (preprocess_input(tf.cast(x, tf.float32)), y), num_parallel_calls=AUTOTUNE)
    return ds.batch(BATCH_SIZE).prefetch(AUTOTUNE)


# %% 5. Model: DenseNet121 + custom head, with block-wise freezing
# Keras DenseNet121 layer naming:
#   conv1_*            stem
#   conv2_block*       dense block 1 (6 layers)     pool2_*  transition 1
#   conv3_block*       dense block 2 (12 layers)    pool3_*  transition 2
#   conv4_block*       dense block 3 (24 layers)    pool4_*  transition 3
#   conv5_block*       dense block 4 (16 layers)
#   bn                 final batch norm
# IMPROVEMENT 6: Phase 1 trains ONLY the new head (random head + pretrained block 4 at lr 1e-3
# damages the pretrained weights). Phase 2 then fine-tunes the entire backbone at a low lr.
PHASE1_TRAINABLE = ()          # head only
PHASE2_TRAINABLE = None        # None = whole backbone trainable


def build_model(hp):
    base = DenseNet121(include_top=False, weights="imagenet", input_shape=(*IMG_SIZE, 3))
    x = layers.GlobalAveragePooling2D(name="gap")(base.output)
    x = layers.Dropout(hp["drop1"], name="drop1")(x)
    x = layers.Dense(hp["dense"], activation="relu", kernel_initializer="glorot_uniform", name="fc256")(x)
    x = layers.Dropout(hp["drop2"], name="drop2")(x)
    # float32 output layer so softmax is numerically safe under mixed precision
    out = layers.Dense(NUM_CLASSES, activation="softmax", kernel_initializer="glorot_uniform",
                       dtype="float32", name="out")(x)
    model = keras.Model(base.input, out, name="densenet121_rice")
    base_names = {l.name for l in base.layers}
    return model, base_names


def set_trainable(model, base_names, prefixes):
    for l in model.layers:
        if l.name in base_names:
            if prefixes is None:
                l.trainable = True
            elif len(prefixes) == 0:
                l.trainable = False
            else:
                l.trainable = l.name.startswith(tuple(prefixes))
        else:
            l.trainable = True  # classification head


def compile_model(model, lr, label_smoothing):
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss=keras.losses.CategoricalCrossentropy(label_smoothing=label_smoothing),
        metrics=["accuracy"],
    )


def make_callbacks(es_patience=None, lr_patience=None):
    es_patience = ES_PATIENCE if es_patience is None else es_patience
    lr_patience = LR_PATIENCE if lr_patience is None else lr_patience
    return [
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=LR_FACTOR, patience=lr_patience, min_lr=1e-7, verbose=1
        ),
        keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=es_patience, restore_best_weights=True, verbose=1
        ),
    ]


def merge_history(h1, h2):
    keys = [k for k in h1.history if k in h2.history]
    return {k: h1.history[k] + h2.history[k] for k in keys}


def get_class_weights(frame):
    if not USE_CLASS_WEIGHTS:
        return None
    w = compute_class_weight("balanced", classes=np.arange(NUM_CLASSES), y=frame["y"].values)
    w = np.sqrt(w)                                   # softened so rare classes are not over-pushed
    return {i: float(v) for i, v in enumerate(w)}


def train_two_phase(train_ds, val_ds, hp=None, class_weight=None, verbose=1,
                    p1_epochs=None, p2_epochs=None, es_patience=None, lr_patience=None):
    hp = HP_DEFAULT if hp is None else hp
    p1_epochs = PHASE1_EPOCHS if p1_epochs is None else p1_epochs
    p2_epochs = PHASE2_EPOCHS if p2_epochs is None else p2_epochs
    model, base_names = build_model(hp)

    # Phase 1 - backbone frozen, train head only
    set_trainable(model, base_names, PHASE1_TRAINABLE)
    compile_model(model, hp["lr1"], hp["ls"])
    h1 = model.fit(train_ds, validation_data=val_ds, epochs=p1_epochs,
                   callbacks=make_callbacks(es_patience, lr_patience),
                   class_weight=class_weight, verbose=verbose)

    # Phase 2 - unfreeze the whole backbone
    set_trainable(model, base_names, PHASE2_TRAINABLE)
    compile_model(model, hp["lr2"], hp["ls"])
    h2 = model.fit(train_ds, validation_data=val_ds, epochs=p2_epochs,
                   callbacks=make_callbacks(es_patience, lr_patience),
                   class_weight=class_weight, verbose=verbose)

    return model, merge_history(h1, h2)


# %% 6. Metrics (Eq. 1-4 of the paper, one-vs-rest per class) -- UNCHANGED
def per_class_metrics(cm):
    cm = cm.astype(np.float64)
    tp = np.diag(cm)
    fn = cm.sum(axis=1) - tp
    fp = cm.sum(axis=0) - tp
    tn = cm.sum() - tp - fn - fp
    eps = 1e-12
    sens = tp / (tp + fn + eps)                        # Eq. 1 (recall)
    spec = tn / (tn + fp + eps)                        # Eq. 2
    acc = (tp + tn) / (tp + tn + fp + fn + eps)        # Eq. 3 (per-class, one-vs-rest)
    prec = tp / (tp + fp + eps)                        # Eq. 4
    f1 = 2 * prec * sens / (prec + sens + eps)
    return sens, spec, prec, f1, acc


def predict_probs(model, ds):
    """Softmax probabilities; with USE_TTA averages original + h-flip + v-flip + both flips."""
    p = model.predict(ds, verbose=0)
    if not USE_TTA:
        return p

    def flipped(fn):
        return ds.map(lambda x, y: (fn(x), y), num_parallel_calls=AUTOTUNE)

    p += model.predict(flipped(lambda x: tf.image.flip_left_right(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(x)), verbose=0)
    p += model.predict(flipped(lambda x: tf.image.flip_up_down(tf.image.flip_left_right(x))), verbose=0)
    return p / 4.0


def evaluate_probs(probs, frame):
    """Same metrics as before, computed from (ensembled) softmax probabilities."""
    y_true = frame["y"].values
    y_pred = probs.argmax(axis=1)
    cm = confusion_matrix(y_true, y_pred, labels=list(range(NUM_CLASSES)))

    sens, spec, prec, f1, acc = per_class_metrics(cm)
    table4 = pd.DataFrame(
        {
            "Sensitivity": sens * 100,
            "Specificity": spec * 100,
            "Precision": prec * 100,
            "F-score": f1 * 100,
            "Accuracy": acc * 100,
        },
        index=PRETTY,
    ).round(2)
    table4.index.name = "Disease Types"

    overall_acc = np.trace(cm) / cm.sum() * 100
    table5 = pd.DataFrame(
        {
            "Sensitivity": [sens.mean() * 100],
            "Specificity": [spec.mean() * 100],
            "Precision": [prec.mean() * 100],
            "F-score": [f1.mean() * 100],
            "Accuracy": [overall_acc],
        },
        index=["Our proposed model"],
    ).round(2)
    return cm, table4, table5, y_true, y_pred


# %% 7. Plots (Fig. 4 and Fig. 5) -- every figure is saved to OUT_DIR (PNG 300 dpi + PDF), then displayed
def save_fig(fig, path):
    path = Path(path)
    fig.savefig(path, dpi=300, bbox_inches="tight")                       # PNG
    fig.savefig(path.with_suffix(".pdf"), bbox_inches="tight")            # vector copy for papers
    print("Saved:", path, "(+ .pdf)")


def plot_curves(hist, path):
    ep = np.arange(1, len(hist["loss"]) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
    ax[0].plot(ep, hist["loss"], "o-", ms=3, label="Training Loss")
    ax[0].plot(ep, hist["val_loss"], "x-", ms=4, label="Validation Loss")
    ax[0].set(title="Training and Validation Loss", xlabel="Epochs", ylabel="Loss")
    ax[1].plot(ep, hist["accuracy"], "o-", ms=3, label="Training Accuracy")
    ax[1].plot(ep, hist["val_accuracy"], "x-", ms=4, label="Validation Accuracy")
    ax[1].set(title="Training and Validation Accuracy", xlabel="Epochs", ylabel="Accuracy")
    for a in ax:
        a.grid(alpha=0.3)
        a.legend()
    plt.tight_layout()
    save_fig(fig, path)
    plt.show()
    plt.close(fig)


def plot_confusion(cm, path):
    cmn = cm / cm.sum(axis=1, keepdims=True)
    fig = plt.figure(figsize=(9, 7))
    sns.heatmap(
        cmn, annot=True, fmt=".2%", cmap="Blues", vmin=0, vmax=1,
        xticklabels=PRETTY, yticklabels=PRETTY, cbar_kws={"label": "Percentage"},
    )
    plt.title("Confusion Matrix for Rice Disease Classification (Normalized by Row)")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    save_fig(fig, path)
    plt.show()
    plt.close(fig)


def plot_tuning(res, path):
    res = res.sort_values("val_acc", ascending=True)
    labels = [f"#{int(t)}" for t in res["trial"]]
    fig, ax = plt.subplots(figsize=(8, 0.45 * len(res) + 1.5))
    ax.barh(labels, res["val_acc"] * 100, color="#4C78A8")
    lo = max(0.0, res["val_acc"].min() * 100 - 2)
    ax.set(xlim=(lo, 100), xlabel="Validation accuracy (%)", ylabel="Trial",
           title="Hyper-parameter search (short runs; #0 = default settings)")
    for y, v in enumerate(res["val_acc"] * 100):
        ax.text(v, y, f" {v:.2f}", va="center", fontsize=8)
    ax.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    save_fig(fig, path)
    plt.show()
    plt.close(fig)


# %% 7b. Hyper-parameter search on the validation split
def run_tuning(train_ds, val_ds, class_weight):
    rng = random.Random(SEED)
    trials, seen = [dict(HP_DEFAULT)], {tuple(sorted(HP_DEFAULT.items()))}
    guard = 0
    while len(trials) < N_TRIALS and guard < 1000:
        guard += 1
        hp = {k: rng.choice(v) for k, v in SEARCH_SPACE.items()}
        key = tuple(sorted(hp.items()))
        if key not in seen:
            seen.add(key)
            trials.append(hp)

    rows = []
    for t, hp in enumerate(trials):
        print(f"\n===== Tuning trial {t + 1}/{len(trials)}: {hp} =====")
        keras.backend.clear_session()
        gc.collect()
        keras.utils.set_random_seed(SEED)
        m, _ = train_two_phase(train_ds, val_ds, hp=hp, class_weight=class_weight, verbose=2,
                               p1_epochs=TUNE_P1_EPOCHS, p2_epochs=TUNE_P2_EPOCHS,
                               es_patience=TUNE_ES_PATIENCE, lr_patience=2)
        v_loss, v_acc = m.evaluate(val_ds, verbose=0)    # weights restored to the best-val_loss epoch
        rows.append({"trial": t, **hp, "val_loss": float(v_loss), "val_acc": float(v_acc)})
        print(f"trial {t}: val_acc={v_acc:.4f}  val_loss={v_loss:.4f}")
        del m

    res = pd.DataFrame(rows)
    res.to_csv(OUT_DIR / "tuning_results.csv", index=False)
    plot_tuning(res, OUT_DIR / "fig_tuning.png")
    best = res.sort_values(["val_acc", "val_loss", "trial"], ascending=[False, True, True]).iloc[0]
    best_hp = {k: type(HP_DEFAULT[k])(best[k]) for k in HP_DEFAULT}
    print("\nBest configuration:", best_hp, f"(val_acc={best['val_acc']:.4f})")
    return best_hp


# %% 8. Main experiment: tune (optional) -> train -> evaluate on the held-out test split
BEST_HP = dict(HP_DEFAULT)
if RUN_MAIN:
    train_ds = make_ds(train_df, training=True)
    val_ds = make_ds(val_df, training=False)
    test_ds = make_ds(test_df, training=False)       # not shuffled, order matches test_df
    cw = get_class_weights(train_df)

    if RUN_TUNING:
        BEST_HP = run_tuning(train_ds, val_ds, cw)
    with open(OUT_DIR / "best_hyperparameters.json", "w") as f:
        json.dump(BEST_HP, f, indent=2)
    print("Hyper-parameters used for the final run:", BEST_HP)

    all_probs, history = [], None
    for k in range(N_ENSEMBLE):
        prob_file = OUT_DIR / f"test_probs_model{k + 1}.npy"
        if RESUME and prob_file.exists():
            print(f"\n[model {k + 1}/{N_ENSEMBLE}] found {prob_file.name}, skipping training")
            all_probs.append(np.load(prob_file))
            continue
        print(f"\n===== Final training: model {k + 1}/{N_ENSEMBLE} =====")
        keras.backend.clear_session()
        gc.collect()
        keras.utils.set_random_seed(SEED + k)
        model, hist_k = train_two_phase(train_ds, val_ds, hp=BEST_HP, class_weight=cw)
        model.save(OUT_DIR / ("densenet121_rice.keras" if N_ENSEMBLE == 1
                              else f"densenet121_rice_model{k + 1}.keras"))
        pd.DataFrame(hist_k).to_csv(OUT_DIR / f"history_model{k + 1}.csv", index=False)
        plot_curves(hist_k, OUT_DIR / f"fig4_curves_model{k + 1}.png")
        p_k = predict_probs(model, test_ds)
        np.save(prob_file, p_k)
        print(f"model {k + 1} alone: test accuracy {(p_k.argmax(1) == test_df['y'].values).mean() * 100:.2f}%")
        all_probs.append(p_k)
        if history is None:
            history = hist_k
        del model

    ens_probs = np.mean(all_probs, axis=0)
    cm, table4, table5, y_true, y_pred = evaluate_probs(ens_probs, test_df)

    print("\nTable 4 - per-class results (%)")
    print(table4)
    print("\nTable 5 - overall results (%)")
    print(table5)

    table4.to_csv(OUT_DIR / "table4_per_class.csv")
    table5.to_csv(OUT_DIR / "table5_overall.csv")
    np.save(OUT_DIR / "confusion_matrix.npy", cm)

    if history is not None:                      # None only if every model was loaded from disk (RESUME)
        pd.DataFrame(history).to_csv(OUT_DIR / "history.csv", index=False)
        plot_curves(history, OUT_DIR / "fig4_curves.png")
    plot_confusion(cm, OUT_DIR / "fig5_confusion_matrix.png")

    print("\nFiles in OUT_DIR:")
    for p in sorted(OUT_DIR.glob("*")):
        if p.suffix in {".png", ".pdf", ".csv", ".json", ".npy", ".keras"}:
            print("  ", p.name)


# %% 9. 5-fold cross-validation (Table 7) and 95% CI / significance (Table 8) -- UNCHANGED logic
def run_cross_validation():
    skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
    rows = []
    for k, (tr_idx, te_idx) in enumerate(skf.split(df, df["y"]), start=1):
        print(f"\n===== Fold {k}/{N_FOLDS} =====")
        keras.backend.clear_session()
        tr_full = df.iloc[tr_idx].reset_index(drop=True)
        te = df.iloc[te_idx].reset_index(drop=True)
        tr, va = split_train_val(tr_full)

        m, _ = train_two_phase(make_ds(tr, True), make_ds(va, False), hp=BEST_HP,
                               class_weight=get_class_weights(tr), verbose=2)

        y_pred = predict_probs(m, make_ds(te, False)).argmax(axis=1)
        y_true = te["y"].values
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)
        rows.append(
            {
                "Fold": f"Fold {k}",
                "Accuracy (%)": accuracy_score(y_true, y_pred) * 100,
                "Precision (%)": p * 100,
                "Recall (%)": r * 100,
                "F1-Score (%)": f * 100,
            }
        )
        print(rows[-1])
        del m

    cv = pd.DataFrame(rows).set_index("Fold")
    metrics = list(cv.columns)

    # Table 7: mean +/- SD (sample SD, ddof=1)
    mean, sd = cv.mean(), cv.std(ddof=1)
    table7 = cv.round(2).copy()
    table7.loc["Mean ± SD"] = [f"{mean[c]:.1f} ± {sd[c]:.2f}" for c in metrics]
    print("\nTable 7 - 5-fold cross-validation")
    print(table7)

    # Table 8: t-based 95% CI over folds + one-sample t-test against a reference baseline
    n = len(cv)
    tcrit = stats.t.ppf(0.975, df=n - 1)
    t8 = []
    for c in metrics:
        half = tcrit * sd[c] / np.sqrt(n)
        _, p = stats.ttest_1samp(cv[c].values, BASELINE_PCT)
        t8.append(
            {
                "Metric": c.replace(" (%)", ""),
                "Mean (%)": round(mean[c], 1),
                "95% CI": f"[{mean[c] - half:.1f}, {mean[c] + half:.1f}]",
                "p-value": "<0.001" if p < 0.001 else f"{p:.3f}",
            }
        )
    table8 = pd.DataFrame(t8).set_index("Metric")
    print(f"\nTable 8 - 95% CI and significance (vs. {BASELINE_PCT}% baseline)")
    print(table8)

    table7.to_csv(OUT_DIR / "table7_cv.csv")
    table8.to_csv(OUT_DIR / "table8_ci.csv")
    return cv, table7, table8


if RUN_CV:
    cv_raw, table7, table8 = run_cross_validation()
```
